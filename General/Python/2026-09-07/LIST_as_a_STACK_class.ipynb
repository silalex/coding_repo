## Stack Implementation using Python Lists

While Python lists can be used as stacks, creating a dedicated Stack class provides better encapsulation and additional functionality:


```python
class Stack:
    # init
    def __init__(self):
        self.stack = []

    # isEmpty: Checks if the stack is empty.
    def isEmpty(self):
        return len(self.stack) == 0

    # Push: Adds a new element on the stack.
    def push(self, a):
        self.stack.append(a)

    # Peek: Returns the top (last) element on the stack.
    def peek(self):
        if self.isEmpty():
            return "Stack is Empty!"
        return self.stack[-1]

    # Pop: Removes and returns the top element from the stack.
    def pop(self):
        if self.isEmpty():
            return "Stack is Empty!"
        return self.stack.pop()

    # Size: Finds the number of elements in the stack.
    def size(self):
        return len(self.stack)
    
# Create a Stack
myStack = Stack()

myStack.push(1)
myStack.push(2)
myStack.push(3)

print("Stack before pop():", myStack.stack)
print("Pop:", myStack.pop())
print("Stack after pop():", myStack.stack)
print("Peek: ", myStack.peek())
print("isEmpty: ", myStack.isEmpty())
print("Size: ", myStack.size())
```
