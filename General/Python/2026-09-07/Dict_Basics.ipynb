# DICT

### Task 1. Sort dict by values

```python
def my_func(d: dict[str, int]) -> dict[str, int]:
    return sorted(d.items(), key=lambda x: x[1])

print(my_func({ "Alice": 95, "David": 88, "Bob": 82, "Charlie": 91}))  # {'Bob': 82, 'David': 88, 'Charlie': 91, 'Alice': 95} 
```

### Task 2. Count elements in a List

```python
def my_func(nums: list[int]) -> dict[int, int]:
    res = {}

    for num in nums:
        res[num] = res.get(num, 0) + 1

    return res

# Test Case
print(my_func([1, 1, 1, 2, 3, 3, 4]))
```

### Task 3. Find all elements appeared more than once

```python
def my_func(nums: list[int]) -> list[int]:
    res = {}

    for num in nums:
        res[num] = res.get(num, 0) + 1

    return [k for k, v in res.items() if v > 1]


print(my_func([1, 2, 2, 3, 3, 3, 4]))
```

### Task 4. Find the most frequent element in the LIST

```python
def my_func(nums: list[int]) -> int:
    res = {}

    for num in nums:
        res[num] = res.get(num, 0) + 1

    return max([v for k, v in res.items()])

print(my_func([1, 2, 2, 3, 3, 3, 4]))   # 3
```

### Task 5. Invert a DICT

```python
def my_func(d: dict[str, int]) -> dict[int, str]:
    return {v: k for k, v in d.items()}

print(my_func({"Alice": 95, "Bob": 82}))
```
