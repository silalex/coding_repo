### Task 1. Case Insensitive Sort
By default the sort() method is case sensitive, resulting in all capital letters being sorted before lower case letters:


````python
# Example: Case sensitive sorting can give an unexpected result
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
````

````python
# Example: Perform a case-insensitive sort of the list
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
````

### Task 2. Customize List Sort Function
Use the keyword argument (key = function) to customize your own SORT function.

The function will return a number that will be used to sort the list (the lowest number first)

````python
# Example: Sort the list based on how close the number is to 50:
def my_func(x: int) -> int:
    return abs (50 - x)

thislist = [100, 50, 60, 70, 80, -1]
thislist.sort(key = my_func)
print(thislist)
````

### Task 3. Remove duples from a List: via set()


````python
def remove_duples(x: list[int]) -> list[int]:
    return list(set(x))

thislist = [0, 0, 0, 3, 4, 7, 7, 8, 9, 9, 10, 10, 10]
print(remove_duples(thislist))
````

### Task 5. Sort strings by length, tie-break alphabetically
Tips: Use a tuple as the sorting key. Python compares tuples element by element (first length, then alphabetical order).

````python
def my_func(x: list[str]) -> list[str]:
    x.sort(key=lambda s: (len(s), s.lower()))
    return x

s = ["q", "aq", "a", "qq", "ab"]
print(my_func(s))
````

### Task 6. Sort numbers: Evens first (ASC), then Odds (ASC)
A tip: A boolean key evaluates False (0) before True (1), splitting the list into two partitions.

````python
def my_func(x: list[int]) -> list[int]:
    return sorted(x, key=lambda x: (x % 2, x))

nums = [7, 2, 9, 4, 1, 8, 6, 3]
print(my_func(nums))
````

### Task 7. Sort numbers: Evens first (ASC), then Odds (DESC)

### Task 11. Remove duples from a sorted List given as a String
Note: Do not use a SET()

````python
def my_func(s: str) -> list[int]:
    # note: map(int, ...) returns a lazy iterator of type <class 'map'>
    nums = list(map(int, s.split(',')))
    res = [nums[0]]

    for n in nums[1:]:
        if n != res[-1]:
            res.append(n)

    return res

s = "1, 1,   2, 2,3,4,4,  5"
print(my_func(s))
````

````python
# minimize the code with LIST comprehension
def my_func_plus(s: str) -> list[int]:
    nums = [int(x) for x in s.split(',')]
    return [n for i, n in enumerate(nums) if 1 == 0 or n != nums[i - 1]]

s = "1, 1,   2, 2,3,4,4,  5"
print(my_func_plus(s))
````

### Leetcode Task 252. Meeting Rooms (check if all intervals not overlapped)
```
You are given an array of meeting times intervals where intervals[i] = [starti, endi].

A person can attend all meetings if no two meeting intervals overlap. Meetings ending at time t and starting at time t do not overlap.

​​​​​​​Return true if a person can attend all meetings. Otherwise, return false.

Example 1:
Input: intervals = [[0,30],[5,10],[15,20]]
Output: false

Example 2:
Input: intervals = [[7,10],[2,4]]
Output: true
```

````python
def my_func(x: list[int]) -> bool:
    x.sort()

    for i in range(1, len(x)):
        if x[i][0] < x[i - 1][1]:
            return False

    return True

intervals = [[0,30],[5,10],[15,20]]
print(my_func(intervals))

intervals = [[7,10],[2,4]]
print(my_func(intervals))
````
