## Lists Sorted Tasks
### Task 1. Case Insensitive Sort
By default the sort() method is case sensitive, resulting in all capital letters being sorted before lower case letters:


```python
# Example: Case sensitive sorting can give an unexpected result
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
```

```python
# Example: Perform a case-insensitive sort of the list
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
```

### Task 2. Customize List Sort Function
Use the keyword argument (key = function) to customize your own SORT function.

The function will return a number that will be used to sort the list (the lowest number first)

```python
# Example: Sort the list based on how close the number is to 50:
def my_func(x: int) -> int:
    return abs (50 - x)

thislist = [100, 50, 60, 70, 80, -1]
thislist.sort(key = my_func)
print(thislist)
```

### Task 3. Remove duples from a List: via set()


```python
def remove_duples(x: list[int]) -> list[int]:
    return list(set(x))

thislist = [0, 0, 0, 3, 4, 7, 7, 8, 9, 9, 10, 10, 10]
print(remove_duples(thislist))
```

### Task 4. Remove duples from a sorted List given as a String

```python
def remove_duples_str(x: str) -> list[int]:
    # return list(map(int, set(x.split(',')))) # bug: ['1', ' 2', ' 2', ' 1']
    return list(set(map(int, x.split(','))))

s = "1, 2, 2, 1"
print(remove_duples_str(s))
```

### Task 5. Sort strings by length, tie-break alphabetically
Tips: Use a tuple as the sorting key. Python compares tuples element by element (first length, then alphabetical order).

```python
def my_func(x: list[str]) -> list[str]:
    x.sort(key=lambda s: (len(s), s.lower()))
    return x

s = ["q", "aq", "a", "qq", "ab"]
print(my_func(s))
```

### Task 6. Sort numbers: Evens first (ASC), then Odds (ASC)
A tip: A boolean key evaluates False ($0$) before True ($1$), splitting the list into two partitions.

```python
def my_func(x: list[int]) -> list[int]:
    return sorted(x, key=lambda x: (x % 2, x))

nums = [7, 2, 9, 4, 1, 8, 6, 3]
print(my_func(nums))
```
